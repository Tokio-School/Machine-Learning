```python
import pandas as pd
import numpy as np
import scipy.cluster.hierarchy as sch
import matplotlib.pyplot as plt

df = pd.read_csv('/content/Dataset_para_Clustering_Jer_rquico (1).csv')

ward = sch.linkage(df, method= 'ward')

plt.figure(figsize=(10,7))
plt.title('Dendograma')
sch.dendrogram(ward)
plt.show()

from scipy.cluster.hierarchy import fcluster

num_cluster = 3
clusters = fcluster(ward, num_cluster, criterion='maxclust')
df['cluster'] = clusters
print(df['cluster'])
```
